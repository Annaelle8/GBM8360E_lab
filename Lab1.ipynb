## EXERCISE 2. 10 points

In practice, the only MRI images available to you from the scanner are a single magnitude image and a single-phase image per acquisition. These images were reconstructed from k-space data acquired on multiple receive coils, and neither the per-coil data nor the combined k-space data can usually be exported from a clinical scanner. Despite this, as an MRI physicist you will often need to infer properties of an acquisition (options used, problems, artifacts) that originate in k-space, using only the magnitude and phase images at your disposal.

This exercise is a refresher on the Fourier transform relationship between k-space and image space, meant to build your intuition for inferring information about an acquisition (which always happens in k-space) from image-space data alone. Note that many of the images shown below are not actual MRI artifacts that you would encounter on a scanner; they are demonstrations of the Fourier transform properties using k-space and MRI images.
